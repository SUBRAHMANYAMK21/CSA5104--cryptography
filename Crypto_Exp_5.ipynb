```python
from math import gcd

def affine_encrypt(text, a, b):
    if gcd(a, 26) != 1:
        return "Invalid key: a must be coprime to 26"

    result = ""
    for ch in text.upper():
        if ch.isalpha():
            p = ord(ch) - ord('A')
            result += chr((a * p + b) % 26 + ord('A'))
        else:
            result += ch

    return result

# Inputs included
plaintext = "HELLO"
a = 5
b = 8

print("Plaintext:", plaintext)
print("a =", a, "b =", b)
print("Ciphertext:", affine_encrypt(plaintext, a, b))

print("Valid values of a:", [x for x in range(1, 26) if gcd(x, 26) == 1])
```
