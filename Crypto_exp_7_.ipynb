```python
from collections import Counter

ciphertext = (
    "53 305 4826 4 806 48 60 85 88 46 88 96 "
    "485 4956 4069 285 8081 48 85 528806"
)

frequency = Counter(ch for ch in ciphertext if ch.isdigit())

print("Ciphertext:", ciphertext)
print("\nCharacter frequencies:")

for char, count in frequency.most_common():
    print(char, ":", count)

print("\nMost frequent characters:")
print(frequency.most_common(5))
```
