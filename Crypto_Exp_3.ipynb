```python
def make_matrix(keyword):
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    result = ""
    for ch in (keyword.upper() + alphabet):
        if ch == "J":
            ch = "I"
        if ch.isalpha() and ch not in result:
            result += ch
    return [list(result[i:i+5]) for i in range(0, 25, 5)]

def prepare_text(text):
    text = ''.join('I' if ch == 'J' else ch for ch in text.upper() if ch.isalpha())
    pairs = []
    i = 0
    while i < len(text):
        a = text[i]
        if i + 1 >= len(text):
            b = 'X'
            i += 1
        elif text[i] == text[i+1]:
            b = 'X'
            i += 1
        else:
            b = text[i+1]
            i += 2
        pairs.append((a, b))
    return pairs

def encrypt(text, keyword):
    matrix = make_matrix(keyword)
    pos = {matrix[r][c]: (r, c) for r in range(5) for c in range(5)}
    result = ""
    for a, b in prepare_text(text):
        r1, c1 = pos[a]
        r2, c2 = pos[b]
        if r1 == r2:
            result += matrix[r1][(c1+1) % 5]
            result += matrix[r2][(c2+1) % 5]
        elif c1 == c2:
            result += matrix[(r1+1) % 5][c1]
            result += matrix[(r2+1) % 5][c2]
        else:
            result += matrix[r1][c2] + matrix[r2][c1]
    return result

# Inputs included
plaintext = "HELLO"
keyword = "MONARCHY"

print("Plaintext:", plaintext)
print("Keyword:", keyword)
print("Ciphertext:", encrypt(plaintext, keyword))
```
