```python
from math import gcd

def decrypt(cipher, a, b):
    inverse = pow(a, -1, 26)
    result = ""
    for ch in cipher:
        if ch.isalpha():
            c = ord(ch.upper()) - 65
            p = (inverse * (c-b)) % 26
            result += chr(p+65)
        else:
            result += ch
    return result

# Inputs included
ciphertext = "BU"
c1, c2 = 1, 20       # B=1, U=20
p1, p2 = 4, 19       # E=4, T=19

for a in range(26):
    if gcd(a, 26) != 1:
        continue
    for b in range(26):
        if (a*p1+b) % 26 == c1 and (a*p2+b) % 26 == c2:
            print("a =", a, "b =", b)
            print("Decrypted:", decrypt(ciphertext, a, b))
```
